## Question 1: Hierarchy of Common Exceptions

### Objective
Create a program that demonstrates the hierarchy of common exceptions such as:
- `ArithmeticException`
- `NullPointerException`
- `ArrayIndexOutOfBoundsException`
- `NumberFormatException`

### Conceptual Background
In object-oriented exception handling systems, exceptions are organized in an inheritance hierarchy rooted in a base class (such as `Exception`). A handler that catches a parent class in the hierarchy will catch any subclass of that parent.

```
Exception
   └── RuntimeException
         ├── ArithmeticException
         ├── NullPointerException
         ├── IndexOutOfBoundsException
         │     └── ArrayIndexOutOfBoundsException
         └── IllegalArgumentException
               └── NumberFormatException
```

In Python, the standard built-in hierarchy has analogous types:
- `ArithmeticError` $\rightarrow$ `ZeroDivisionError`, `OverflowError`, `FloatingPointError`
- `TypeError` / `AttributeError` $\rightarrow$ operations or method calls on `None`
- `LookupError` $\rightarrow$ `IndexError` (accessing list/array indices out of bounds)
- `ValueError` $\rightarrow$ attempting to convert invalid strings to numeric types (`int("abc")`)

Below, we define a complete object-oriented class hierarchy mirroring these standard exceptions, demonstrate polymorphic exception handling (catching child exceptions via their parent types), and also demonstrate Python's native built-in exception hierarchy.


````python
# Q1: Exception Hierarchy Implementation

class RuntimeException(Exception):
    """Base class for unchecked/runtime exceptions."""
    pass

class ArithmeticException(RuntimeException):
    """Raised when an exceptional arithmetic condition has occurred (e.g., division by zero)."""
    pass

class NullPointerException(RuntimeException):
    """Raised when an application attempts to use None where an object instance is required."""
    pass

class IndexOutOfBoundsException(RuntimeException):
    """Raised to indicate that an index of some sort is out of range."""
    pass

class ArrayIndexOutOfBoundsException(IndexOutOfBoundsException):
    """Raised to indicate that an array has been accessed with an illegal index."""
    pass

class IllegalArgumentException(RuntimeException):
    """Raised to indicate an illegal or inappropriate argument passed to a method."""
    pass

class NumberFormatException(IllegalArgumentException):
    """Raised when attempting to convert a string to a numeric type with an invalid format."""
    pass

print("=== Custom Exception Hierarchy Defined Successfully ===")
print("ArrayIndexOutOfBoundsException -> IndexOutOfBoundsException -> RuntimeException -> Exception")
print("NumberFormatException -> IllegalArgumentException -> RuntimeException -> Exception")
print("ArithmeticException -> RuntimeException -> Exception")
print("NullPointerException -> RuntimeException -> Exception")
````

````python
# Q1: Demonstrating Hierarchy and Polymorphic Exception Handling

def trigger_arithmetic_exception(a, b):
    if b == 0:
        raise ArithmeticException(f"ArithmeticException: Division by zero attempted ({a} / {b}).")
    return a / b

def trigger_null_pointer_exception(obj):
    if obj is None:
        raise NullPointerException("NullPointerException: Attempted to invoke method or access field on None object.")
    return len(obj)

def trigger_array_bounds_exception(lst, index):
    if index < 0 or index >= len(lst):
        raise ArrayIndexOutOfBoundsException(
            f"ArrayIndexOutOfBoundsException: Index {index} out of bounds for array of length {len(lst)}."
        )
    return lst[index]

def trigger_number_format_exception(text):
    text = text.strip()
    # Check if text is a valid integer string (with optional leading sign)
    if not (text.isdigit() or (text.startswith(('-', '+')) and text[1:].isdigit())):
        raise NumberFormatException(f"NumberFormatException: For input string: '{text}'")
    return int(text)

print("--- 1. Catching ArrayIndexOutOfBoundsException via Parent Classes ---")
test_array = [10, 20, 30]

# Level 1: Caught by immediate parent IndexOutOfBoundsException
try:
    trigger_array_bounds_exception(test_array, 5)
except IndexOutOfBoundsException as e:
    print(f"[Caught by IndexOutOfBoundsException]: {e}")

# Level 2: Caught by grandparent RuntimeException
try:
    trigger_array_bounds_exception(test_array, -1)
except RuntimeException as e:
    print(f"[Caught by RuntimeException]: {e}")

print("\n--- 2. Catching NumberFormatException via Parent Classes ---")
# Level 1: Caught by immediate parent IllegalArgumentException
try:
    trigger_number_format_exception("1024_invalid")
except IllegalArgumentException as e:
    print(f"[Caught by IllegalArgumentException]: {e}")

# Level 2: Caught by grandparent RuntimeException
try:
    trigger_number_format_exception("3.1415_not_int")
except RuntimeException as e:
    print(f"[Caught by RuntimeException]: {e}")

print("\n--- 3. Catching ArithmeticException via RuntimeException ---")
try:
    trigger_arithmetic_exception(50, 0)
except RuntimeException as e:
    print(f"[Caught by RuntimeException]: {e}")

print("\n--- 4. Catching NullPointerException via RuntimeException ---")
try:
    trigger_null_pointer_exception(None)
except RuntimeException as e:
    print(f"[Caught by RuntimeException]: {e}")

print("\n--- 5. Inheritance Verification (issubclass checks) ---")
print("issubclass(ArrayIndexOutOfBoundsException, IndexOutOfBoundsException):", 
      issubclass(ArrayIndexOutOfBoundsException, IndexOutOfBoundsException))
print("issubclass(ArrayIndexOutOfBoundsException, RuntimeException):         ", 
      issubclass(ArrayIndexOutOfBoundsException, RuntimeException))
print("issubclass(NumberFormatException, IllegalArgumentException):           ", 
      issubclass(NumberFormatException, IllegalArgumentException))
print("issubclass(NumberFormatException, RuntimeException):                   ", 
      issubclass(NumberFormatException, RuntimeException))
print("issubclass(ArithmeticException, RuntimeException):                     ", 
      issubclass(ArithmeticException, RuntimeException))
print("issubclass(NullPointerException, RuntimeException):                    ", 
      issubclass(NullPointerException, RuntimeException))
````

## Question 2: Nested try-catch Blocks & Exception Propagation

### Objective
Write a program containing nested `try-catch` (in Python, `try-except`) blocks. Generate different exceptions in the inner and outer blocks and demonstrate how exception propagation works.

### Key Concepts Demonstrated:
1. **Inner Exception Caught Locally:** The inner block generates an exception that matches an inner `except` handler. The inner block handles it, and outer block execution continues normally.
2. **Inner Exception Propagated to Outer Block:** The inner block generates an exception that does NOT match any inner handler. The exception propagates outward to the enclosing `try-except` block.
3. **Inner Exception Re-raised:** The inner block catches an exception, performs partial handling (e.g. logging), and explicitly re-raises it (`raise`) to let the outer block handle it.
4. **Outer Block Exception:** An exception occurs directly in the outer block (before or after the inner block) and is intercepted by the outer handler.


````python
# Q2: Nested try-except Blocks and Exception Propagation Demo

def demonstrate_nested_propagation(scenario_id: int):
    print(f"\n================ Scenario {scenario_id} ================")
    
    if scenario_id == 1:
        print("[Goal]: Inner block generates ZeroDivisionError -> Handled by inner block -> Outer block finishes normally.")
    elif scenario_id == 2:
        print("[Goal]: Inner block generates IndexError -> NOT handled by inner block -> Propagates to Outer block.")
    elif scenario_id == 3:
        print("[Goal]: Inner block generates ValueError -> Caught by inner block -> Re-raised to Outer block.")
    elif scenario_id == 4:
        print("[Goal]: Outer block itself generates TypeError before/after inner block -> Outer handler catches it.")

    try:
        print("  -> Entering OUTER try block")
        
        if scenario_id == 4:
            print("  -> Outer block: Simulating TypeError (outer level failure)...")
            result = "String" + 500  # TypeError
            
        # NESTED INNER TRY BLOCK
        try:
            print("     -> Entering INNER try block")
            
            if scenario_id == 1:
                print("     -> Inner block: Triggering ZeroDivisionError (10 / 0)...")
                val = 10 / 0
            elif scenario_id == 2:
                print("     -> Inner block: Triggering IndexError (accessing index 99)...")
                nums = [1, 2, 3]
                val = nums[99]  # Unhandled by inner except ZeroDivisionError
            elif scenario_id == 3:
                print("     -> Inner block: Triggering ValueError (int('abc'))...")
                val = int("invalid_number_abc")
                
            print("     -> Inner block completed successfully without exceptions.")
            
        except ZeroDivisionError as inner_err:
            print(f"     [INNER HANDLER - ZeroDivisionError]: Caught locally! Message: {inner_err}")
            print("     [INNER HANDLER]: Recovered successfully inside inner scope.")
            
        except ValueError as inner_err:
            print(f"     [INNER HANDLER - ValueError]: Caught! Performing logging and re-raising...")
            print("     [INNER HANDLER]: Re-raising exception to outer block using 'raise'.")
            raise  # Re-raise to propagate outwards
            
        finally:
            print("     [INNER finally]: Inner cleanup executed.")
            
        print("  -> Resuming OUTER try block after inner block.")
        print("  -> Outer block operations completed successfully.")
        
    except IndexError as outer_err:
        print(f"  [OUTER HANDLER - IndexError]: Caught propagated exception from inner block! Message: {outer_err}")
        
    except ValueError as outer_err:
        print(f"  [OUTER HANDLER - ValueError]: Caught re-raised exception from inner block! Message: {outer_err}")
        
    except TypeError as outer_err:
        print(f"  [OUTER HANDLER - TypeError]: Caught exception generated directly in outer block! Message: {outer_err}")
        
    except Exception as outer_err:
        print(f"  [OUTER HANDLER - General Exception]: Caught {type(outer_err).__name__}: {outer_err}")
        
    finally:
        print("  [OUTER finally]: Outer cleanup executed (Always guaranteed).")

# Execute all scenarios
demonstrate_nested_propagation(1)
demonstrate_nested_propagation(2)
demonstrate_nested_propagation(3)
demonstrate_nested_propagation(4)
````

## Question 3: Calculator with Specialized Exception Handlers

### Objective
Develop a calculator that accepts two numbers and an operator (`+`, `-`, `*`, `/`). It should be able to handle:
- **Division by zero** (`ZeroDivisionError`)
- **Invalid numeric input** (`ValueError`)
- **Invalid operators** (Custom `InvalidOperatorException`)

Use separate exception handlers for different errors, along with `else` and `finally` blocks.


````python
# Q3: Calculator Implementation

class InvalidOperatorException(Exception):
    """Custom exception raised when an unsupported operator is supplied."""
    def __init__(self, operator):
        self.operator = operator
        super().__init__(f"InvalidOperatorException: Operator '{operator}' is not supported. Supported operators are: +, -, *, /")


def robust_calculator(num1_raw, num2_raw, operator):
    """
    Performs arithmetic operations (+, -, *, /) with separate exception handlers.
    """
    print(f"\n--- Calculating: {num1_raw} {operator} {num2_raw} ---")
    result = None
    
    try:
        # 1. Parse numeric inputs - may raise ValueError
        try:
            n1 = float(num1_raw)
            n2 = float(num2_raw)
        except (ValueError, TypeError) as e:
            raise ValueError(f"Invalid numeric input: Could not convert '{num1_raw}' or '{num2_raw}' to a valid number.") from e
        
        # 2. Validate operator - may raise InvalidOperatorException
        valid_operators = ['+', '-', '*', '/']
        if operator not in valid_operators:
            raise InvalidOperatorException(operator)
            
        # 3. Perform computation - may raise ZeroDivisionError
        if operator == '+':
            result = n1 + n2
        elif operator == '-':
            result = n1 - n2
        elif operator == '*':
            result = n1 * n2
        elif operator == '/':
            if n2 == 0:
                raise ZeroDivisionError(f"Division by zero error: Cannot divide {n1} by zero.")
            result = n1 / n2

    except ZeroDivisionError as zde:
        print(f"  [ERROR HANDLER: ZeroDivisionError] -> {zde}")
        
    except ValueError as ve:
        print(f"  [ERROR HANDLER: ValueError] -> {ve}")
        
    except InvalidOperatorException as ioe:
        print(f"  [ERROR HANDLER: InvalidOperatorException] -> {ioe}")
        
    except Exception as unexpected:
        print(f"  [ERROR HANDLER: Unexpected Error] -> {type(unexpected).__name__}: {unexpected}")
        
    else:
        # Executes only when NO exception was raised
        # If integer equivalent, display cleanly
        formatted_res = int(result) if result.is_integer() else round(result, 4)
        print(f"  [SUCCESS]: {n1} {operator} {n2} = {formatted_res}")
        return result
        
    finally:
        # Always executes
        print("  [AUDIT]: Calculator operation cycle finished.")


# Test Suite for Calculator
print("=== Running Comprehensive Calculator Tests ===")

# Test Case 1: Valid operations
robust_calculator("15", "25", "+")
robust_calculator("100", "37.5", "-")
robust_calculator("12.5", "4", "*")
robust_calculator("100", "8", "/")

# Test Case 2: Division by zero
robust_calculator("50", "0", "/")
robust_calculator("-12.4", "0.0", "/")

# Test Case 3: Invalid numeric input
robust_calculator("abc", "10", "+")
robust_calculator("25", "xyz123", "*")
robust_calculator("12.34.56", "2", "/")

# Test Case 4: Invalid operators
robust_calculator("10", "3", "%")
robust_calculator("2", "8", "^")
robust_calculator("15", "5", "add")
````

## Question 4: Student Result-Processing Program

### Objective
Create a student result-processing program that accepts marks for several subjects.
- Marks must be between **0 and 100**.
- Create a custom `InvalidMarksException`.
- Throw the exception when invalid marks (negative, $> 100$, or non-numeric) are entered.
- Calculate **total**, **percentage**, and **grade** only when **all input is valid**.
- Display a meaningful error message upon failure.


````python
# Q4: Student Result Processing Implementation

class InvalidMarksException(Exception):
    """Custom exception thrown when invalid marks are encountered."""
    def __init__(self, subject, marks, reason):
        self.subject = subject
        self.marks = marks
        self.reason = reason
        self.message = f"InvalidMarksException: Subject '{subject}' received invalid marks [{marks}]. Reason: {reason}"
        super().__init__(self.message)


def compute_grade(percentage):
    """Determines letter grade based on percentage."""
    if percentage >= 90:
        return "A+ (Outstanding)"
    elif percentage >= 80:
        return "A (Excellent)"
    elif percentage >= 70:
        return "B (Very Good)"
    elif percentage >= 60:
        return "C (Good)"
    elif percentage >= 50:
        return "D (Pass)"
    else:
        return "F (Fail)"


def process_student_results(student_name, roll_no, marks_dict):
    """
    Validates subject marks, ensuring all marks are between 0 and 100.
    Calculates total, percentage, and grade only when all inputs are valid.
    """
    print(f"\n{'='*55}")
    print(f"Processing Result for: {student_name} (Roll No: {roll_no})")
    print(f"{'='*55}")
    
    try:
        # Phase 1: STRICT VALIDATION OF ALL SUBJECTS BEFORE ANY COMPUTATION
        print("Validating entered marks for all subjects...")
        validated_marks = {}
        
        for subject, score in marks_dict.items():
            # Check 1: Numeric type validation
            if isinstance(score, bool) or not isinstance(score, (int, float)):
                raise InvalidMarksException(
                    subject=subject,
                    marks=score,
                    reason=f"Marks must be a valid number, got {type(score).__name__} ('{score}')"
                )
            
            # Check 2: Range validation (0 <= score <= 100)
            if score < 0:
                raise InvalidMarksException(
                    subject=subject,
                    marks=score,
                    reason=f"Marks cannot be negative. Must be >= 0."
                )
            if score > 100:
                raise InvalidMarksException(
                    subject=subject,
                    marks=score,
                    reason=f"Marks cannot exceed maximum limit of 100."
                )
                
            validated_marks[subject] = float(score)
            print(f"  [OK] {subject:18s}: {score}")

        # Phase 2: COMPUTATION (Only executed if ALL marks are strictly valid)
        print("\nAll inputs validated successfully! Computing result...")
        total_marks_obtained = sum(validated_marks.values())
        max_possible_marks = len(validated_marks) * 100
        percentage = (total_marks_obtained / max_possible_marks) * 100
        grade = compute_grade(percentage)
        
        # Display Result Card
        print(f"\n------------------ SCORECARD ------------------")
        print(f" Student Name : {student_name}")
        print(f" Roll Number  : {roll_no}")
        print(f" Total Marks  : {total_marks_obtained:.2f} / {max_possible_marks}")
        print(f" Percentage   : {percentage:.2f}%")
        print(f" Final Grade  : {grade}")
        print(f" Status       : {'PASSED' if percentage >= 50 else 'FAILED'}")
        print(f"-----------------------------------------------")
        
        return {
            "name": student_name,
            "roll_no": roll_no,
            "total": total_marks_obtained,
            "percentage": percentage,
            "grade": grade,
            "status": "PASSED" if percentage >= 50 else "FAILED"
        }

    except InvalidMarksException as ime:
        print(f"\n[RESULT PROCESSING FAILED]:\n  -> {ime}")
        print("  -> Action aborted: Total, percentage, and grade were NOT computed.")
        return None
    except Exception as e:
        print(f"\n[UNEXPECTED ERROR]: {type(e).__name__}: {e}")
        return None


# Test Cases
# 1. Valid Input
marks_valid = {
    "Mathematics": 95,
    "Data Structures": 88,
    "Operating Systems": 92,
    "Cognitive Computing": 96,
    "Technical English": 85
}
process_student_results("Rishabh Dutt", "1024170393", marks_valid)

# 2. Negative Marks Error
marks_negative = {
    "Mathematics": 85,
    "Data Structures": -15,   # INVALID
    "Operating Systems": 78
}
process_student_results("Alice Sharma", "1024170301", marks_negative)

# 3. Marks > 100 Error
marks_overflow = {
    "Cognitive Computing": 105, # INVALID
    "Machine Learning": 90,
    "Cyber Security": 80
}
process_student_results("Bob Verma", "1024170302", marks_overflow)

# 4. Non-numeric Marks Error
marks_string = {
    "Mathematics": 78,
    "Operating Systems": "absent",  # INVALID
    "Database Systems": 82
}
process_student_results("Charlie Singh", "1024170303", marks_string)
````

## Question 5: Banking Application with Exception Propagation

### Objective
Implement a simple banking application supporting deposit and withdrawal.
- Create custom exceptions:
  - `InsufficientBalanceException`
  - `InvalidAmountException`
  - `AccountNotFoundException`
- Prevent withdrawal when the balance is insufficient.
- Prevent zero or negative deposits/withdrawals.
- **Demonstrate exception propagation from one method to another:**
  - `BankAccount.withdraw()` raises an exception.
  - `BankService.transfer()` calls `withdraw()` without catching it, allowing it to bubble up to the `TransactionManager` / caller.
- Display the account balance after successful transactions.
- Handle invalid account numbers with `AccountNotFoundException`.


````python
# Q5: Banking Application Implementation

class BankingException(Exception):
    """Base exception for banking application."""
    pass

class InsufficientBalanceException(BankingException):
    """Raised when attempting to withdraw more funds than available in balance."""
    def __init__(self, acc_no, current_balance, attempted_amount):
        self.acc_no = acc_no
        self.current_balance = current_balance
        self.attempted_amount = attempted_amount
        super().__init__(
            f"InsufficientBalanceException: Account [{acc_no}] has balance ₹{current_balance:,.2f}, "
            f"which cannot satisfy withdrawal of ₹{attempted_amount:,.2f}."
        )

class InvalidAmountException(BankingException):
    """Raised when transaction amount is non-positive or invalid."""
    def __init__(self, operation_type, amount):
        self.operation_type = operation_type
        self.amount = amount
        super().__init__(
            f"InvalidAmountException: Cannot perform {operation_type} with non-positive amount: ₹{amount}. "
            f"Transaction amount must be strictly greater than zero."
        )

class AccountNotFoundException(BankingException):
    """Raised when an account number does not exist in the bank records."""
    def __init__(self, acc_no):
        self.acc_no = acc_no
        super().__init__(f"AccountNotFoundException: Account number [{acc_no}] not found in bank database.")


class BankAccount:
    """Represents a customer bank account with deposit and withdrawal capabilities."""
    def __init__(self, account_number: str, holder_name: str, initial_balance: float = 0.0):
        self.account_number = account_number
        self.holder_name = holder_name
        self.balance = float(initial_balance)

    def deposit(self, amount: float):
        """Validates amount and deposits funds."""
        if amount <= 0:
            raise InvalidAmountException("DEPOSIT", amount)
        self.balance += amount
        return self.balance

    def withdraw(self, amount: float):
        """Validates amount and checks sufficient funds before withdrawing."""
        if amount <= 0:
            raise InvalidAmountException("WITHDRAWAL", amount)
        if amount > self.balance:
            raise InsufficientBalanceException(self.account_number, self.balance, amount)
        self.balance -= amount
        return self.balance


class BankService:
    """Bank management service demonstrating exception propagation."""
    def __init__(self):
        self.accounts = {}

    def add_account(self, account: BankAccount):
        self.accounts[account.account_number] = account

    def get_account(self, account_number: str) -> BankAccount:
        """Looks up an account; raises AccountNotFoundException if absent."""
        if account_number not in self.accounts:
            raise AccountNotFoundException(account_number)
        return self.accounts[account_number]

    def process_deposit(self, account_number: str, amount: float):
        """Low-level propagation: calls get_account and deposit."""
        acc = self.get_account(account_number)  # May propagate AccountNotFoundException
        new_balance = acc.deposit(amount)       # May propagate InvalidAmountException
        return acc, new_balance

    def process_withdrawal(self, account_number: str, amount: float):
        """Low-level propagation: calls get_account and withdraw."""
        acc = self.get_account(account_number)  # May propagate AccountNotFoundException
        new_balance = acc.withdraw(amount)      # May propagate InsufficientBalanceException/InvalidAmountException
        return acc, new_balance

    def transfer_funds(self, from_acc_no: str, to_acc_no: str, amount: float):
        """
        Demonstrates Multi-Level Exception Propagation:
        This method does NOT catch exceptions locally.
        Any exception in get_account() or withdraw() propagates up to caller.
        """
        print(f"  [BankService.transfer_funds] Initiating transfer: ₹{amount:,.2f} from {from_acc_no} to {to_acc_no}")
        sender = self.get_account(from_acc_no)
        receiver = self.get_account(to_acc_no)
        
        # Withdraw from sender (may raise InsufficientBalanceException or InvalidAmountException)
        sender.withdraw(amount)
        # Deposit into receiver
        receiver.deposit(amount)
        print(f"  [BankService.transfer_funds] Transfer internal state updated successfully.")
        return sender.balance, receiver.balance


def transaction_manager(action_description, action_func):
    """Top-level transaction controller that intercepts all propagated banking exceptions."""
    print(f"\n--- Executing: {action_description} ---")
    try:
        result = action_func()
        print(f"  [TRANSACTION SUCCESSFUL]")
        return result
    except AccountNotFoundException as anfe:
        print(f"  [FAILED - AccountNotFoundException]: {anfe}")
    except InsufficientBalanceException as ibe:
        print(f"  [FAILED - InsufficientBalanceException]: {ibe}")
    except InvalidAmountException as iae:
        print(f"  [FAILED - InvalidAmountException]: {iae}")
    except BankingException as be:
        print(f"  [FAILED - BankingException]: {be}")
    except Exception as e:
        print(f"  [UNEXPECTED SYSTEM ERROR]: {type(e).__name__}: {e}")
    finally:
        print("  [AUDIT]: Banking transaction record processed.")


# Setup Bank Accounts
bank = BankService()
acc1 = BankAccount("ACC-101", "Rishabh Dutt", 10000.0)
acc2 = BankAccount("ACC-102", "Priya Patel", 5000.0)
bank.add_account(acc1)
bank.add_account(acc2)

# Test 1: Successful Deposit
def test_valid_deposit():
    acc, bal = bank.process_deposit("ACC-101", 3000.0)
    print(f"  Account {acc.account_number} ({acc.holder_name}) new balance: ₹{bal:,.2f}")
transaction_manager("Deposit ₹3,000 into ACC-101", test_valid_deposit)

# Test 2: Successful Withdrawal
def test_valid_withdrawal():
    acc, bal = bank.process_withdrawal("ACC-101", 4000.0)
    print(f"  Account {acc.account_number} ({acc.holder_name}) new balance: ₹{bal:,.2f}")
transaction_manager("Withdraw ₹4,000 from ACC-101", test_valid_withdrawal)

# Test 3: Prevent Zero or Negative Deposit
transaction_manager("Deposit ₹0 into ACC-101", lambda: bank.process_deposit("ACC-101", 0.0))
transaction_manager("Deposit ₹-500 into ACC-101", lambda: bank.process_deposit("ACC-101", -500.0))

# Test 4: Prevent Zero or Negative Withdrawal
transaction_manager("Withdraw ₹-200 from ACC-101", lambda: bank.process_withdrawal("ACC-101", -200.0))

# Test 5: Prevent Withdrawal when Balance is Insufficient
transaction_manager("Withdraw ₹50,000 from ACC-101", lambda: bank.process_withdrawal("ACC-101", 50000.0))

# Test 6: AccountNotFoundException for invalid account number
transaction_manager("Access non-existent account ACC-999", lambda: bank.process_deposit("ACC-999", 1000.0))

# Test 7: Exception Propagation during Multi-Account Transfer
print("\n=== Demonstrating Multi-Method Exception Propagation during Transfer ===")
# 7a: Insufficient balance transfer (propagates from BankAccount.withdraw -> BankService.transfer_funds -> transaction_manager)
transaction_manager(
    "Transfer ₹25,000 from ACC-101 to ACC-102 (exceeds balance)",
    lambda: bank.transfer_funds("ACC-101", "ACC-102", 25000.0)
)

# 7b: Invalid recipient account (propagates from BankService.get_account -> transfer_funds -> transaction_manager)
transaction_manager(
    "Transfer ₹1,000 from ACC-101 to ACC-999 (invalid recipient)",
    lambda: bank.transfer_funds("ACC-101", "ACC-999", 1000.0)
)

# 7c: Successful Transfer
def test_successful_transfer():
    b1, b2 = bank.transfer_funds("ACC-101", "ACC-102", 2000.0)
    print(f"  ACC-101 updated balance: ₹{b1:,.2f}")
    print(f"  ACC-102 updated balance: ₹{b2:,.2f}")
transaction_manager("Transfer ₹2,000 from ACC-101 to ACC-102", test_successful_transfer)
````

## Question 6: Login System with Exception-Based Validation

### Objective
Develop a login system with exception-based validation:
- Create custom exceptions:
  - `InvalidUsernameException`
  - `InvalidPasswordException`
  - `AccountLockedException`
- Allow only a fixed number of failed login attempts (e.g., 3).
- **Lock the account** after the maximum attempts are reached.
- Use `try-catch-finally` appropriately.
- Display different messages for different failures.


````python
# Q6: Login Authentication System Implementation
import time

class LoginValidationException(Exception):
    """Base exception for login validation errors."""
    pass

class InvalidUsernameException(LoginValidationException):
    """Raised when the username is unrecognized or formatted illegally."""
    def __init__(self, username, message="Username does not exist in our directory."):
        self.username = username
        super().__init__(f"InvalidUsernameException: User '{username}' was not found. {message}")

class InvalidPasswordException(LoginValidationException):
    """Raised when the entered password does not match the registered credential."""
    def __init__(self, username, attempts_left):
        self.username = username
        self.attempts_left = attempts_left
        super().__init__(
            f"InvalidPasswordException: Incorrect password for '{username}'. "
            f"Failed attempt registered. Attempts remaining before lockout: {attempts_left}"
        )

class AccountLockedException(LoginValidationException):
    """Raised when a login attempt is made against a locked user account."""
    def __init__(self, username, max_attempts):
        self.username = username
        self.max_attempts = max_attempts
        super().__init__(
            f"AccountLockedException: Account for '{username}' is LOCKED due to {max_attempts} "
            f"consecutive failed login attempts. Please contact administrative support."
        )


class UserAccountRecord:
    """Maintains state for a user account."""
    def __init__(self, username, password, max_attempts=3):
        self.username = username
        self.password = password
        self.failed_attempts = 0
        self.max_attempts = max_attempts
        self.is_locked = False

    def reset_failures(self):
        self.failed_attempts = 0


class AuthenticationService:
    """Service managing user credentials, validations, and account locks."""
    def __init__(self):
        self.user_database = {}

    def register_user(self, username, password, max_attempts=3):
        self.user_database[username] = UserAccountRecord(username, password, max_attempts)

    def login(self, username, password):
        """
        Validates username and password; raises specific exceptions on failure.
        """
        # 1. Check Username
        if username not in self.user_database:
            raise InvalidUsernameException(username)

        user = self.user_database[username]

        # 2. Check if Account is already Locked
        if user.is_locked:
            raise AccountLockedException(username, user.max_attempts)

        # 3. Check Password
        if user.password != password:
            user.failed_attempts += 1
            attempts_remaining = user.max_attempts - user.failed_attempts
            
            if user.failed_attempts >= user.max_attempts:
                user.is_locked = True
                raise AccountLockedException(username, user.max_attempts)
            else:
                raise InvalidPasswordException(username, attempts_remaining)

        # 4. Successful Authentication
        user.reset_failures()
        return f"SESSION_TOKEN_AUTH_OK_{username.upper()}"


def attempt_login(auth_system: AuthenticationService, username_input: str, password_input: str):
    """
    Executes a login attempt wrapped in a try-except-finally structure.
    Demonstrates guaranteed cleanup and audit logging in the 'finally' block.
    """
    masked_pw = '*' * len(password_input) if password_input else '<EMPTY>'
    print(f"\n--- Login Attempt for User: '{username_input}' with Password: '{masked_pw}' ---")
    
    session_token = None
    login_status = "UNKNOWN"
    
    try:
        session_token = auth_system.login(username_input, password_input)
        login_status = "SUCCESS"
        print(f"  [LOGIN GRANTED]: Welcome back, {username_input}!")
        print(f"  [SESSION]: Active session token -> {session_token}")
        
    except InvalidUsernameException as ue:
        login_status = "FAILED_INVALID_USERNAME"
        print(f"  [ERROR - Invalid Username]: {ue}")
        
    except InvalidPasswordException as pe:
        login_status = "FAILED_INVALID_PASSWORD"
        print(f"  [ERROR - Invalid Password]: {pe}")
        
    except AccountLockedException as le:
        login_status = "ACCOUNT_LOCKED"
        print(f"  [SECURITY ALERT - Account Locked]: {le}")
        
    except Exception as unexpected:
        login_status = "SYSTEM_ERROR"
        print(f"  [UNEXPECTED ERROR]: {type(unexpected).__name__}: {unexpected}")
        
    finally:
        # GUARANTEED AUDIT LOGGING IN FINALLY BLOCK
        user_record = auth_system.user_database.get(username_input)
        attempts = user_record.failed_attempts if user_record else "N/A"
        locked = user_record.is_locked if user_record else "N/A"
        print(f"  [FINALLY BLOCK AUDIT]:")
        print(f"    - Event Status     : {login_status}")
        print(f"    - Failed Attempts  : {attempts}")
        print(f"    - Locked Status    : {locked}")
        print(f"    - Clean-up         : Connection credentials cleared from memory buffer.")


# Setup Authentication System
auth_service = AuthenticationService()
auth_service.register_user("rishabh", "SecurePass@2026", max_attempts=3)

print("=== Starting Login Validation Demonstration ===")

# Test 1: Successful Login
attempt_login(auth_service, "rishabh", "SecurePass@2026")

# Test 2: Unregistered Username
attempt_login(auth_service, "ghost_user", "AnyPassword123")

# Test 3: Failed Login Attempt 1 (Incorrect Password)
attempt_login(auth_service, "rishabh", "WrongPassword1")

# Test 4: Failed Login Attempt 2 (Incorrect Password)
attempt_login(auth_service, "rishabh", "WrongPassword2")

# Test 5: Failed Login Attempt 3 (Triggers Account Lockout)
attempt_login(auth_service, "rishabh", "WrongPassword3")

# Test 6: Subsequent Login Attempt on Locked Account (Even with CORRECT password)
attempt_login(auth_service, "rishabh", "SecurePass@2026")
````

## Question 7: Shopping/Cart Application with Complete Exception Hierarchy

### Objective
Design a small shopping/cart application that uses a complete exception hierarchy:

```
ApplicationException
├── ProductException
│   ├── ProductNotFoundException
│   └── OutOfStockException
├── PaymentException
│   ├── InvalidPaymentException
│   └── InsufficientFundsException
└── OrderException
    └── EmptyCartException
```

Implement operations like:
- Search products
- Add / remove products to/from cart
- Checkout and payment processing
- Appropriate exception handlers for both specific and group-level exceptions
- Programmatic debugging demonstration using **variable inspection** and **breakpoint inspection** at points of error.


````python
# Q7: Exception Hierarchy Definition

class ApplicationException(Exception):
    """Root exception for all application-level errors in the e-commerce system."""
    def __init__(self, message, error_code="APP_ERR"):
        self.message = message
        self.error_code = error_code
        super().__init__(f"[{self.error_code}] {self.message}")

# --- Product Exceptions Branch ---
class ProductException(ApplicationException):
    """Base exception for all product-related catalog and inventory failures."""
    def __init__(self, message, error_code="PROD_ERR"):
        super().__init__(message, error_code)

class ProductNotFoundException(ProductException):
    """Raised when a requested product ID or name does not exist in catalog."""
    def __init__(self, product_id):
        self.product_id = product_id
        super().__init__(f"Product with ID '{product_id}' was not found in the catalog.", "PROD_NOT_FOUND")

class OutOfStockException(ProductException):
    """Raised when requested quantity exceeds available inventory stock."""
    def __init__(self, product_id, product_name, requested_qty, available_stock):
        self.product_id = product_id
        self.product_name = product_name
        self.requested_qty = requested_qty
        self.available_stock = available_stock
        super().__init__(
            f"Cannot fulfill '{product_name}' (ID: {product_id}): requested {requested_qty} units, "
            f"but only {available_stock} units in stock.",
            "OUT_OF_STOCK"
        )

# --- Payment Exceptions Branch ---
class PaymentException(ApplicationException):
    """Base exception for payment processing errors."""
    def __init__(self, message, error_code="PAY_ERR"):
        super().__init__(message, error_code)

class InvalidPaymentException(PaymentException):
    """Raised when payment method, card number, or gateway details are malformed."""
    def __init__(self, reason):
        self.reason = reason
        super().__init__(f"Invalid payment information provided: {reason}", "INVALID_PAYMENT")

class InsufficientFundsException(PaymentException):
    """Raised when customer's account balance cannot cover the order total."""
    def __init__(self, current_funds, required_amount):
        self.current_funds = current_funds
        self.required_amount = required_amount
        super().__init__(
            f"Insufficient funds: Available balance ₹{current_funds:,.2f} is less than required ₹{required_amount:,.2f}.",
            "INSUFFICIENT_FUNDS"
        )

# --- Order Exceptions Branch ---
class OrderException(ApplicationException):
    """Base exception for order processing errors."""
    def __init__(self, message, error_code="ORDER_ERR"):
        super().__init__(message, error_code)

class EmptyCartException(OrderException):
    """Raised when attempting to place an order or checkout with an empty cart."""
    def __init__(self):
        super().__init__("Cannot proceed to checkout: The shopping cart is empty.", "EMPTY_CART")

print("=== ApplicationException Hierarchy Successfully Established ===")
````

````python
# Q7: Domain Models and Operations Implementation

class Product:
    def __init__(self, product_id: str, name: str, price: float, stock: int):
        self.product_id = product_id
        self.name = name
        self.price = float(price)
        self.stock = int(stock)

    def __repr__(self):
        return f"<Product {self.product_id}: {self.name} | ₹{self.price:,.2f} | Stock: {self.stock}>"


class ShoppingCart:
    def __init__(self):
        # Maps product_id -> {'product': Product, 'quantity': int}
        self.items = {}

    def is_empty(self):
        return len(self.items) == 0

    def add(self, product: Product, quantity: int):
        if product.product_id in self.items:
            self.items[product.product_id]['quantity'] += quantity
        else:
            self.items[product.product_id] = {'product': product, 'quantity': quantity}

    def remove(self, product_id: str, quantity: int = None):
        if product_id not in self.items:
            raise ProductNotFoundException(product_id)
        
        if quantity is None or quantity >= self.items[product_id]['quantity']:
            del self.items[product_id]
        else:
            self.items[product_id]['quantity'] -= quantity

    def calculate_total(self):
        return sum(item['product'].price * item['quantity'] for item in self.items.values())

    def clear(self):
        self.items.clear()

    def get_summary(self):
        lines = []
        for item in self.items.values():
            prod = item['product']
            qty = item['quantity']
            lines.append(f"  * {prod.name} (x{qty}) @ ₹{prod.price:,.2f} = ₹{prod.price * qty:,.2f}")
        return lines


class PaymentProcessor:
    """Handles payment verification and execution."""
    VALID_METHODS = ["CREDIT_CARD", "DEBIT_CARD", "UPI", "NET_BANKING"]

    @staticmethod
    def process_payment(payment_method: str, card_or_upi_id: str, amount: float, customer_wallet_balance: float):
        # 1. Validate payment method
        if payment_method.upper() not in PaymentProcessor.VALID_METHODS:
            raise InvalidPaymentException(f"Unsupported payment method '{payment_method}'. Supported methods: {PaymentProcessor.VALID_METHODS}")

        # 2. Validate card number / UPI ID format
        clean_id = card_or_upi_id.replace(" ", "").replace("-", "")
        if payment_method.upper() in ["CREDIT_CARD", "DEBIT_CARD"]:
            if not (clean_id.isdigit() and len(clean_id) == 16):
                raise InvalidPaymentException(f"Card number '{card_or_upi_id}' is invalid. Must be a 16-digit numeric card number.")
        elif payment_method.upper() == "UPI":
            if "@" not in clean_id:
                raise InvalidPaymentException(f"UPI ID '{card_or_upi_id}' is invalid. Format must be 'username@provider'.")

        # 3. Validate customer funds
        if customer_wallet_balance < amount:
            raise InsufficientFundsException(customer_wallet_balance, amount)

        # Successful deduction
        new_balance = customer_wallet_balance - amount
        return f"TXN_GATEWAY_{abs(hash(clean_id)) % 100000000}", new_balance


class ECommerceApp:
    """Coordinates catalog, cart, orders, and payments."""
    def __init__(self):
        self.catalog = {}
        self.cart = ShoppingCart()

    def add_catalog_product(self, product: Product):
        self.catalog[product.product_id] = product

    def search_product(self, query: str) -> Product:
        """Search product by exact ID or substring in name; raises ProductNotFoundException if absent."""
        query_lower = query.lower().strip()
        # Search by exact product_id
        if query in self.catalog:
            return self.catalog[query]
        # Search by name match
        for prod in self.catalog.values():
            if query_lower in prod.name.lower():
                return prod
        raise ProductNotFoundException(query)

    def add_to_cart(self, product_id: str, quantity: int):
        """Searches product and verifies stock before adding to cart."""
        product = self.search_product(product_id)  # May raise ProductNotFoundException
        
        # Determine total requested quantity including any already in cart
        current_in_cart = self.cart.items.get(product_id, {}).get('quantity', 0)
        total_requested = current_in_cart + quantity
        
        if total_requested > product.stock:
            raise OutOfStockException(product.product_id, product.name, total_requested, product.stock)
            
        self.cart.add(product, quantity)
        print(f"  [Cart Updated]: Added {quantity}x '{product.name}' to cart.")

    def remove_from_cart(self, product_id: str, quantity: int = None):
        self.cart.remove(product_id, quantity)
        print(f"  [Cart Updated]: Removed product '{product_id}' from cart.")

    def checkout(self, customer_name: str, payment_method: str, payment_id: str, customer_balance: float):
        """Performs order checkout, verifying cart contents, payment, and inventory update."""
        print(f"\n--- Checkout for {customer_name} ---")
        
        # 1. Verify Cart is NOT Empty
        if self.cart.is_empty():
            raise EmptyCartException()
            
        order_total = self.cart.calculate_total()
        print(f"  Order Items:")
        for line in self.cart.get_summary():
            print(f"  {line}")
        print(f"  Order Total: ₹{order_total:,.2f}")
        
        # 2. Process Payment (may raise PaymentException subclasses)
        txn_id, updated_balance = PaymentProcessor.process_payment(
            payment_method, payment_id, order_total, customer_balance
        )
        
        # 3. Deduct Inventory Stock
        for item in self.cart.items.values():
            p = item['product']
            q = item['quantity']
            p.stock -= q
            
        # 4. Clear Cart
        self.cart.clear()
        
        print(f"  [CHECKOUT COMPLETE]: Transaction ID: {txn_id}")
        print(f"  Remaining customer wallet balance: ₹{updated_balance:,.2f}")
        return txn_id
````

````python
# Q7: Debugging with Breakpoints and Variable Inspection

print("=== Debugging Demonstration: Variable Inspection & Stack Inspection ===")

def debug_inspector(func, *args, **kwargs):
    """
    Simulates debugger breakpoint and variable inspection on exception:
    Inspects local frame variables, arguments, and object state at the point of failure.
    """
    try:
        return func(*args, **kwargs)
    except ApplicationException as e:
        print(f"\n[DEBUG BREAKPOINT TRIGGERED]: Intercepted {type(e).__name__}!")
        print(f"  * Error Code        : {e.error_code}")
        print(f"  * Error Message     : {e.message}")
        print(f"  * Exception Attributes:")
        for attr, val in e.__dict__.items():
            if attr not in ['message', 'error_code']:
                print(f"      - {attr}: {val}")
        
        # Variable Inspection: Inspecting application & cart state
        print(f"  * Application State Inspection:")
        print(f"      - Cart Item Count : {len(store.cart.items)}")
        print(f"      - Cart Items      : {list(store.cart.items.keys())}")
        print(f"      - Catalog Size    : {len(store.catalog)} products")
        if isinstance(e, OutOfStockException):
            prod = store.catalog.get(e.product_id)
            print(f"      - Current Catalog Stock for {e.product_id}: {prod.stock if prod else 'N/A'}")
        print("  [DEBUGGER]: State inspection complete. Program execution safely resumed.\n")

# Trigger breakpoint and state inspection on OutOfStock
debug_inspector(store.add_to_cart, "P104", 50)

# Trigger breakpoint and state inspection on ProductNotFound
debug_inspector(store.add_to_cart, "P888", 2)
````
